# Метрика

Перед решением любой задачи важно понимать, как будет оцениваться ваше решение. В данном случае мы используем стандартную для задачи классификации метрику ROC-AUC. Ее можно вычислить, используя только предсказанные вероятности и истинные классы без конкретного порога классификации + она раотает даже если классы в данных сильно несбалансированны (примеров одного класса в десятки раз больше примеров длугого). Именно поэтому она очень удобна для соревнований.

Посчитать ее легко:


```python
from sklearn.metrics import roc_auc_score

y_true = [
    0,
    1,
    1,
    0,
    1
]

y_predictions = [
    0.1,
    0.9,
    0.4,
    0.6,
    0.61
]

roc_auc_score(y_true, y_predictions)
```

```python
from sklearn.model_selection import train_test_split
train_data, val_data = train_test_split(data, train_size=0.9, random_state=42)
```

```python
val_data.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
ordinal_encoder = OrdinalEncoder()

train_data['HasMultiplePhoneNumbers'] = train_data['HasMultiplePhoneNumbers'].replace('No phone service', 'No')
val_data['HasMultiplePhoneNumbers'] = val_data['HasMultiplePhoneNumbers'].replace('No phone service', 'No')

internet_cols = ['HasOnlineSecurityService', 'HasOnlineBackup', 'HasDeviceProtection', 'HasTechSupportAccess', 'HasOnlineTV', 'HasMovieSubscription']
train_data[internet_cols] = train_data[internet_cols].replace('No internet service', 'No')
val_data[internet_cols] = val_data[internet_cols].replace('No internet service', 'No')

mask = train_data.select_dtypes('object').nunique().eq(2)
binary_cat_cols = mask[mask].index

train_data[binary_cat_cols] = ordinal_encoder.fit_transform(train_data[binary_cat_cols])
val_data[binary_cat_cols] = ordinal_encoder.transform(val_data[binary_cat_cols])
train_data[binary_cat_cols]
```

```python
train_data['TotalSpent'].isna().sum()
```

```python
train_data['TotalSpent'] = train_data['TotalSpent'].replace(' ', np.nan).astype(float)
TotalSpent_median = train_data['TotalSpent'].median()
train_data['TotalSpent'] = train_data['TotalSpent'].fillna(TotalSpent_median)
val_data['TotalSpent'] = val_data['TotalSpent'].replace(' ', np.nan).astype(float)
val_data['TotalSpent'] = val_data['TotalSpent'].fillna(TotalSpent_median)
```

```python
data.select_dtypes('object')
```

```python
from sklearn.preprocessing import OneHotEncoder
onehot_encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False,
    dtype=int
)
binary_cols = train_data.select_dtypes('object').columns
train_encoded = onehot_encoder.fit_transform(train_data[binary_cols])
val_encoded = onehot_encoder.transform(val_data[binary_cols])


encoded_columns = onehot_encoder.get_feature_names_out(binary_cols)

train_encoded = pd.DataFrame(
    train_encoded,
    columns=encoded_columns,
    index=train_data.index
)

val_encoded = pd.DataFrame(
    val_encoded,
    columns=encoded_columns,
    index=val_data.index
)

train_data = pd.concat(
    [
        train_data.drop(columns=binary_cols),
        train_encoded
    ],
    axis=1
)

val_data = pd.concat(
    [
        val_data.drop(columns=binary_cols),
        val_encoded
    ],
    axis=1
)
```

```python
from sklearn.preprocessing import  StandardScaler
numeric_cols = train_data.select_dtypes(include='number').columns

numeric_cols = [
    col for col in numeric_cols
    if train_data[col].nunique() > 2
]

scaler = StandardScaler()

train_data[numeric_cols] = scaler.fit_transform(
    train_data[numeric_cols]
)

val_data[numeric_cols] = scaler.transform(
    val_data[numeric_cols]
)
```

```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
from sklearn.metrics import roc_auc_score, roc_curve, f1_score, accuracy_score, precision_score, recall_score
```

```python
X_train = train_data.drop('Churn',axis=1)
y_train = train_data['Churn']
X_val = val_data.drop('Churn',axis=1)
y_val = val_data['Churn']
```

```python
pipeline = make_pipeline(
    LogisticRegression(
        max_iter=2000,
        class_weight='balanced',
        random_state=42
    )
)

pipeline.fit(X_train, y_train)
```

```python
threshold = 0.4
preds_proba = pipeline.predict_proba(X_val)[:,1]
preds = (preds_proba >= threshold).astype(int)
```

```python
metrics = pd.Series({
    'Accuracy': accuracy_score(np.array(y_val), preds), 'precision': precision_score(np.array(y_val), preds),
    'Recall': recall_score(np.array(y_val), preds), 'F1': f1_score(np.array(y_val), preds), 'roc_auc': roc_auc_score(np.array(y_val), preds_proba)
    }, name='Score').rename_axis('Metric')
```

```python
metrics
```

```python
fpr, tpr, thresholds = roc_curve(y_val, preds_proba)
roc_auc = roc_auc_score(y_val, preds_proba)

plt.figure(figsize=(7, 6))

plt.plot(
    fpr,
    tpr,
    label=f'ROC-AUC = {roc_auc:.3f}'
)

# Линия случайного классификатора
plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--',
    label='Случайный классификатор'
)

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC-кривая')
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
